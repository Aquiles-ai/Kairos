```python
import torch
from transformers import AutoModelForCausalLM, AutoProcessor, TextStreamer
from transformers.image_utils import load_image

model = AutoModelForCausalLM.from_pretrained("Aquiles-ai/Kairos-Alig-30k", trust_remote_code=True, dtype=torch.bfloat16).to("cuda")
processor = AutoProcessor.from_pretrained("Aquiles-ai/Kairos-Alig-30k", trust_remote_code=True)
streamer = TextStreamer(processor, skip_prompt=True, skip_special_tokens=True)

prompt = "Hey, how are you?"
message = [{"role": "user", "content": prompt}]

enc = processor.apply_chat_template(message,
  tokenize=True, return_dict=True, return_tensors="pt",
  add_generation_prompt=True,
).to(model.device)

out = model.generate(
  **enc, max_new_tokens=1024, streamer=streamer
)
```

```python
question = "What's in this image?"
question2 = "Is there a bicycle in the scene? If so, describe the scene, its location, and who is riding it (obviously only if someone is using it; otherwise, don't mention it)."
question3 = "Tell me anything you notice, whatever it is."
url = "https://hips.hearstapps.com/hmg-prod/images/road-bike-tips-0255-65aaae8a59db1.jpg?crop=0.668xw:1.00xh;0.207xw,0&resize=640:*"
image = load_image(url)

print(f"Question 1: {question}")
print(f"Question 2: {question2}")
print(f"Question 3: {question3}")
print(f"The model is being shown an image of a woman on a bike")

message_img = [{"role": "user", "content": [
    {"type": "image", "image": image},
    {"type": "text", "text": question},
]}]

message_img2 = [{"role": "user", "content": [
    {"type": "image", "image": image},
    {"type": "text", "text": question2},
]}]

message_img3 = [{"role": "user", "content": [
    {"type": "image", "image": image},
    {"type": "text", "text": question3},
]}]

enc_im = processor.apply_chat_template(message_img,
  tokenize=True, return_dict=True, return_tensors="pt",
  add_generation_prompt=True
).to(model.device)


enc_im2 = processor.apply_chat_template(message_img2,
  tokenize=True, return_dict=True, return_tensors="pt",
  add_generation_prompt=True
).to(model.device)

enc_im3 = processor.apply_chat_template(message_img3,
  tokenize=True, return_dict=True, return_tensors="pt",
  add_generation_prompt=True
).to(model.device)

print("Output for question 1")
out_img = model.generate(
  **enc_im, max_new_tokens=1024, streamer=streamer
)

print("Output for question 2")
out_img2 = model.generate(
  **enc_im2, max_new_tokens=1024, streamer=streamer
)

print("Output for question 3")
out_img3 = model.generate(
  **enc_im3, max_new_tokens=1024, streamer=streamer
)
```
